# NYC 311 Service Requests: End-to-End Data Cleaning Pipeline

## 1. Load Data

```python
import pandas as pd
import numpy as np

url = "https://data.cityofnewyork.us/resource/erm2-nwe9.csv?$limit=50000"
df = pd.read_csv(url, low_memory=False)
print(f"Dataset loaded with shape: {df.shape}")
```

## 2. Temporal Data Formatting

```python
df['created_date'] = pd.to_datetime(df['created_date'], errors='coerce')
```

## 3. General Categorical & Numeric Imputation

```python
numeric_cols = df.select_dtypes(include=['number']).columns
df[numeric_cols] = df[numeric_cols].fillna(0)

exclude_cols = ['created_date', 'closed_date', 'due_date']
non_numeric_cols = [col for col in df.columns if col not in numeric_cols and col not in exclude_cols]
df[non_numeric_cols] = df[non_numeric_cols].fillna('unknown')
```

## 4. Cleaning Specialized Date Fields

```python
df['due_date'] = df['due_date'].astype(str).replace(['NaT', '<NA>', 'nan', 'NaN'], 'unknown')
df['due_date'] = df['due_date'].fillna('unknown')

df['closed_date'] = df['closed_date'].astype(str).replace(['NaT', '<NA>', 'nan', 'NaN'], 'still open')
df['closed_date'] = df['closed_date'].fillna('still open')
```

## 5. Final Validation Report

```python
missing_count = df.isnull().sum().sum()
print(f"Total missing values remaining: {missing_count}")
if missing_count == 0:
    print("🎉 Dataset is 100% clean and ready for analytical insights!")
```
