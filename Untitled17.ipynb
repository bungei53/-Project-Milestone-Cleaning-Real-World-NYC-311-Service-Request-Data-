```python
# =========================================================================
# SECTION: OUTLIER DETECTION HELPER FUNCTION (IQR METHOD)
# =========================================================================
# This function uses the Interquartile Range (IQR) method to calculate standard
# statistical boundaries (1.5 * IQR) and flags any values falling outside them.

def detect_outlier(df, columns):
  Q1 = df[columns].quantile(0.25)
  Q3 = df[columns].quantile(0.75)

  IQR = Q3 - Q1

  lower_bound = Q1 - (1.5 * IQR)
  upper_bound = Q3 + (1.5 * IQR)

  outlier = df[
      (df[columns] < lower_bound) | (df[columns] > upper_bound)
  ]
  return {
      'columns': columns,
      'Q1': Q1,
      'Q3': Q3,
      'IQR': IQR,
      'lower_bound': lower_bound,
      'upper_bound': upper_bound,
      'outlier_count': len(outlier),
  }
```

```python
# =========================================================================
# SECTION: GENERATING OUTLIER SUMMARY REPORT
# =========================================================================
# Here we map our outlier helper function across all numerical columns
# in our dataset to obtain a structured dataset profile.

outlier_report = pd.DataFrame(
    [detect_outlier(df, col) for col in numeric_col.columns]
)
display(outlier_report)
```
